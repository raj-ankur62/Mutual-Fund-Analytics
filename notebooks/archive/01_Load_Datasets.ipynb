```python
import pandas as pd
import os
```

```python
folder_path = "../data/raw"

csv_files = [
    file for file in os.listdir(folder_path)
    if file.endswith(".csv") and file[:2].isdigit()
]

csv_files
```

```python
for file in csv_files:
    print("=" * 80)
    print(f"Dataset: {file}")

    file_path = os.path.join(folder_path, file)

    df = pd.read_csv(file_path)

    print("\nShape:")
    print(df.shape)

    print("\nData Types:")
    print(df.dtypes)

    print("\nFirst 5 Rows:")
    print(df.head())

    print("\nMissing Values:")
    print(df.isnull().sum())

    print("\nDuplicate Rows:")
    print(df.duplicated().sum())

    print("=" * 80)
```

```python
import pandas as pd

fund_master = pd.read_csv(
    "../data/raw/01_fund_master.csv"
)

nav_history = pd.read_csv(
    "../data/raw/02_nav_history.csv"
)

print("Fund Master Shape:", fund_master.shape)
print("NAV History Shape:", nav_history.shape)
```

```python
fund_codes = set(fund_master["amfi_code"])

nav_codes = set(nav_history["amfi_code"])
```

```python
missing_codes = fund_codes - nav_codes

print("AMFI Codes missing in NAV History:")
print(missing_codes)

print("\nTotal Missing Codes:")
print(len(missing_codes))
```

```python
extra_codes = nav_codes - fund_codes

print("Extra AMFI Codes in NAV History:")
print(extra_codes)

print("\nTotal Extra Codes:")
print(len(extra_codes))
```

```python
quality_summary = {
    "Total Fund Master Schemes": fund_master["amfi_code"].nunique(),
    "Total NAV History Schemes": nav_history["amfi_code"].nunique(),
    "Missing AMFI Codes": len(missing_codes),
    "Extra AMFI Codes": len(extra_codes),
    "NAV Records": len(nav_history)
}

quality_summary
```
