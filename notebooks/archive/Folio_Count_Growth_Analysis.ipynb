```python
## Markdown
# Folio Count Growth Analysis

## Objective
- Analyze folio count growth from Jan 2022 to Dec 2025.
- Create a line chart.
- Highlight key milestones.
```

```python
# Import Libraries
import pandas as pd
import plotly.express as px
```

```python
# Load Dataset
df = pd.read_csv("../data/processed/06_industry_folio_count_clean.csv")
```

```python
# Check Dataset
print(df.shape)

print(df.columns)

df.head()
```

```python
# Convert Month
df["month"] = pd.to_datetime(df["month"])

print(df.dtypes)
```

```python
# Create Line Chart
fig = px.line(
    df,
    x="month",
    y="total_folios_crore",
    markers=True,
    title="Industry Folio Count Growth (Jan 2022 – Dec 2025)",
    labels={
        "month": "Month",
        "total_folios_crore": "Total Folios (Crore)"
    }
)
```

```python
## Mark Key Milestones
# First milestone (Jan 2022)
fig.add_annotation(
    x=df.iloc[0]["month"],
    y=df.iloc[0]["total_folios_crore"],
    text="13.26 Cr<br>Jan 2022",
    showarrow=True,
    arrowhead=2,
    arrowcolor="green",
    ax=-40,
    ay=-50
)

# Final milestone (Dec 2025)
fig.add_annotation(
    x=df.iloc[-1]["month"],
    y=df.iloc[-1]["total_folios_crore"],
    text="26.12 Cr<br>Dec 2025",
    showarrow=True,
    arrowhead=2,
    arrowcolor="red",
    ax=-70,
    ay=-60
)
```

```python
# Improve Layout
fig.update_layout(
    template="plotly_white",
    hovermode="x unified",
    height=650,
    width=1300,
    xaxis_title="Month",
    yaxis_title="Total Folios (Crore)"
)
```

```python
# Show Chart
fig.show()
```

```python
# Save Report
fig.write_html("../reports/10_folio_count_growth.html")
```
