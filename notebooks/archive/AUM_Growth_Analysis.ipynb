```python
# AUM Growth Analysis (2022–2025)

### Objective
- Compare AUM growth of all fund houses.
- Create grouped bar chart using Seaborn.
- Highlight SBI dominance (₹12.5 Lakh Crore).
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use("ggplot")
```

```python
# Load dataset
df = pd.read_csv("../data/processed/03_aum_by_fund_house_clean.csv")
```

```python
# check dataset
df.head()
```

```python
print(df.columns)
print(df.shape)
```

```python
## Convert date & extract year
df["date"] = pd.to_datetime(df["date"])

df["year"] = df["date"].dt.year

df.head()
```

```python
Filter(2022-2025)
df = df[
    (df["year"] >= 2022) &
    (df["year"] <= 2025)
]

print(df.shape)
```

```python
# print(df[df["fund_house"].str.contains("SBI", case=False, na=False)])
```

```python
#Highlight SBI (₹12.5 Lakh Crore)
plt.figure(figsize=(15,8))

ax = sns.barplot(
    data=df,
    x="year",
    y="aum_lakh_crore",
    hue="fund_house"
)

plt.title("AUM Growth by Fund House (2022–2025)")
plt.xlabel("Year")
plt.ylabel("AUM (₹ Lakh Crore)")

plt.annotate(
    "SBI Dominance\n₹12.5 Lakh Cr",
    xy=(3, 12.5),
    xytext=(2.7, 12.8),
    arrowprops=dict(arrowstyle="->", color="red", lw=2),
    fontsize=11,
    color="red",
    fontweight="bold"
)

plt.legend(title="Fund House", bbox_to_anchor=(1.02,1), loc="upper left")

plt.tight_layout()
plt.show()
```
