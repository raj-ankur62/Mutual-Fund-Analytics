```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("../data/processed/07_scheme_performance_clean.csv")

# Top 10 funds by 1-year return
top_funds = df.sort_values("return_1yr_pct", ascending=False).head(10)

# Plot
plt.figure(figsize=(12,6))
plt.bar(top_funds["scheme_name"], top_funds["return_1yr_pct"])

plt.xticks(rotation=90)
plt.xlabel("Scheme Name")
plt.ylabel("1 Year Return (%)")
plt.title("Top 10 Funds by 1 Year Return")

plt.tight_layout()
plt.savefig("../reports/13_fund_performance_comparison.png")
plt.show()
```

```python
df = pd.read_csv("../data/processed/08_investor_transactions_clean.csv")

transaction_count = df["transaction_type"].value_counts()

plt.figure(figsize=(6,4))

plt.bar(
    transaction_count.index,
    transaction_count.values
)

plt.title("Transaction Type Distribution")
plt.xlabel("Transaction Type")
plt.ylabel("Count")

plt.savefig("../reports/15_transaction_type_analysis.png")

plt.show()
```

```python


# Load dataset
df = pd.read_csv("../data/processed/03_aum_by_fund_house_clean.csv")

# Agar multiple dates hain to latest date ka data lo
latest_df = df[df["date"] == df["date"].max()]

plt.figure(figsize=(12,6))

plt.bar(latest_df["fund_house"], latest_df["aum_crore"])

plt.xticks(rotation=90)
plt.title("AUM Distribution by Fund House")
plt.xlabel("Fund House")
plt.ylabel("AUM (Crore)")

plt.tight_layout()

plt.savefig("../reports/16_aum_distribution.png")

plt.show()
```
