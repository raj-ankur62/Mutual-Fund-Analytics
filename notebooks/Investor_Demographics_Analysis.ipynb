```python
## Markdown
# Investor Demographics Analysis

## Objective
- Analyze investor age group distribution.
- Compare SIP amount by age group using a box plot.
- Visualize gender distribution.
```

```python
## Import Libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use("ggplot")
```

```python
# Load Dataset
df = pd.read_csv("../data/processed/08_investor_transactions_clean.csv")
```

```python
# Check Dataset
print(df.shape)

print(df.columns)

df.head()
```

```python
# Age Group Distribution (Pie Chart)
age_counts = df["age_group"].value_counts()

plt.figure(figsize=(8,8))

plt.pie(
    age_counts,
    labels=age_counts.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Investor Age Group Distribution")

plt.savefig(
    "../reports/05_age_group_distribution.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# SIP Amount by Age Group (Box Plot)
sip_df = df[df["transaction_type"] == "SIP"]

plt.figure(figsize=(10,6))

sns.boxplot(
    data=sip_df,
    x="age_group",
    y="amount_inr"
)

plt.title("SIP Amount Distribution by Age Group")
plt.xlabel("Age Group")
plt.ylabel("SIP Amount (₹)")

plt.tight_layout()

plt.savefig(
    "../reports/06_sip_amount_boxplot.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# Gender Distribution (Bar Chart)
plt.figure(figsize=(6,5))

sns.countplot(
    data=df,
    x="gender"
)

plt.title("Investor Gender Distribution")
plt.xlabel("Gender")
plt.ylabel("Number of Investors")

plt.tight_layout()

plt.savefig(
    "../reports/07_gender_distribution.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
