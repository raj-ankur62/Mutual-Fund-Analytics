```python
# SIP Inflow Trend Analysis (Jan 2022 – Dec 2025)

## Objective
# - Analyze monthly SIP inflows.
# - Plot monthly SIP trend using Plotly.
# - Highlight the all-time high SIP inflow of ₹31,002 Crore in December 2025.
```

```python
# import libraries
import pandas as pd
import plotly.express as px
```

```python
# Load dataset
df = pd.read_csv("../data/processed/04_monthly_sip_inflows_clean.csv")
```

```python
# check dataset
print(df.shape)

print(df.columns)

df.head()
```

```python
# Convert Month to datetime
df["month"] = pd.to_datetime(df["month"])

df.head()
```

```python
# Plot Monthly SIP Trend
fig = px.line(
    df,
    x="month",
    y="sip_inflow_crore",
    markers=True,
    title="Monthly SIP Inflow Trend (Jan 2022 – Dec 2025)",
    labels={
        "month": "Month",
        "sip_inflow_crore": "SIP Inflow (₹ Crore)"
    }
)
```

```python
# Find highest SIP inflow

max_row = df.loc[df["sip_inflow_crore"].idxmax()]

fig.add_annotation(
    x=max_row["month"],
    y=max_row["sip_inflow_crore"],
    text=f"All-Time High<br>₹{max_row['sip_inflow_crore']:,} Cr",
    showarrow=True,
    arrowhead=2,
    arrowsize=1.2,
    arrowwidth=2,
    arrowcolor="red",
    ax=-90,
    ay=-60,
    font=dict(
        color="red",
        size=12
    ),
    bgcolor="white",
    bordercolor="red",
    borderwidth=1
)
```

```python
# Improve Layout
fig.update_layout(
    template="plotly_white",
    hovermode="x unified",
    height=650,
    width=1300,
    xaxis_title="Month",
    yaxis_title="SIP Inflow (₹ Crore)"
)
```

```python
# Show Chart
fig.show()
```

```python
fig.write_html("../reports/04_sip_inflow_trend.html")
```
